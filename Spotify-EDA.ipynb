```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('Spotify_dataset.csv')


columns_irrelevant = ['track_id', 'mode','key','time_signature','Unnamed: 0']
variables=['instrumentalness','liveness','duration_ms','energy','loudness','speechiness','acousticness','valence','tempo',]
genres = ['pop','jazz','classical','hip-hop']
df.drop(columns_irrelevant, axis = 1)
```

```python
variance_genre = df.groupby('track_genre')[variables].std()

plot_var_genre = variance_genre.loc[genres]

plot_var_genre
```

```python
plot_var_genre = plot_var_genre.reset_index()

plt.figure(figsize=(10,6))

sns.barplot(data=plot_var_genre, x='track_genre',y='energy')

plt.title("Variance of Energy by Genre")
plt.show()
```

```python
mean_genre = df.groupby('track_genre')[['popularity']].mean()

plot_pop_genre = mean_genre.loc[genres]

plot_pop_genre
```

```python
plot_pop_genre = plot_pop_genre.reset_index()

plt.figure(figsize=(10,6))

sns.barplot(data=plot_pop_genre, x='track_genre',y='popularity')

plt.title("Mean Popularity by Genre")
plt.show()
```
